```python
company_sql = """
INSERT INTO companies (
    company_id, name, description, state, country,city, zip_code, address, url
)
VALUES (%s, %s, %s, %s, %s, %s, %s, %s, %s)
ON CONFLICT (company_id) DO NOTHING;
"""

try:
    cur.executemany(
        company_sql,
        companies.itertuples(index=False, name=None)
    )
    conn.commit()
    print("✅ Columns inserted successfully")
except Exception as e:
    conn.rollback()
    print("❌ Insert failed:", e)
```

```python
jobs['listed_at'] = pd.to_datetime(jobs['listed_time'], unit='ms', errors='coerce')
jobs = jobs[jobs['listed_at'].notnull()]

int_cols = ['job_id', 'company_id', 'applies', 'original_listed_time', 
            'expiry', 'closed_time', 'sponsored', 'max_salary', 'med_salary', 'min_salary', 'normalized_salary']

conn = psycopg2.connect(
    dbname='linkedin_jobs',
    user='postgres',
    password='postgres',
    host='localhost',
    port='5432'
)
cur = conn.cursor()

def safe_bigint_convert(x):
    if pd.isnull(x):
        return None
    try:
        val = float(x)
        if np.isinf(val) or np.isnan(val):
            return None
        val = int(val)
        if val < -9223372036854775808 or val > 9223372036854775807:
            return None
        return val
    except (ValueError, OverflowError, TypeError):
        return None

insert_sql = """
INSERT INTO jobs(
    job_id, company_id, company_name, title, description,
    max_salary, med_salary, min_salary, pay_period,
    formatted_work_type, location, applies, original_listed_time,
    remote_allowed, job_posting_url, application_url,
    application_type, expiry, closed_time,
    formatted_experience_level, skills_desc, listed_at, posting_domain,
    sponsored, work_type, currency, compensation_type, normalized_salary,
    zip_code, fips
)
VALUES (
    %s, %s, %s, %s, %s,
    %s, %s, %s, %s,
    %s, %s, %s, %s,
    %s, %s, %s,
    %s, %s, %s,
    %s, %s, %s, %s,
    %s, %s, %s, %s, %s,
    %s, %s
)
ON CONFLICT (job_id, listed_at) DO NOTHING;
"""

commit_interval = 20000 

for idx, row in jobs.iterrows():
    try:
        int_values = {col: safe_bigint_convert(row.get(col)) for col in int_cols}

        values = (
            int_values.get('job_id'),
            int_values.get('company_id'),
            row.get('company_name'),
            row.get('title'),
            row.get('description'),
            int_values.get('max_salary'),
            int_values.get('med_salary'),
            int_values.get('min_salary'),
            row.get('pay_period'),
            row.get('formatted_work_type'),
            row.get('location'),
            int_values.get('applies'),
            int_values.get('original_listed_time'),
            bool(row.get('remote_allowed')) if pd.notnull(row.get('remote_allowed')) else None,
            row.get('job_posting_url'),
            row.get('application_url'),
            row.get('application_type'),
            int_values.get('expiry'),
            int_values.get('closed_time'),
            row.get('formatted_experience_level'),
            row.get('skills_desc'),
            row.get('listed_at'),
            row.get('posting_domain'),
            int_values.get('sponsored'),
            row.get('work_type'),
            row.get('currency'),
            row.get('compensation_type'),
            int_values.get('normalized_salary'),
            row.get('zip_code'),
            row.get('fips')
        )
        
        cur.execute(insert_sql, values)
        
        if (idx + 1) % commit_interval == 0:
            conn.commit()
            print(f"Inserted {idx + 1} rows...")

    except Exception as e:
        print(f"Error on row {idx}: {e}")
        conn.rollback()

conn.commit()
cur.close()
conn.close()
print("All rows processed!")
```

```python
conn = psycopg2.connect(
    dbname="linkedin_jobs",
    user="postgres",
    password="postgres",
    host="localhost",
    port="5432"
)
cur = conn.cursor()
print("✅ Connected to PostgreSQL")

# --- Skills ---
skills_clean = skills[['skill_abr','skill_name']].copy()
skills_clean = skills_clean.where(pd.notnull(skills_clean), None)

# --- Industries ---
industries_clean = industries[['industry_id','industry_name']].copy()
industries_clean['industry_id'] = industries_clean['industry_id'].astype(int)
industries_clean = industries_clean.where(pd.notnull(industries_clean), None)


def insert_table(df, sql):
    """Safe batch insert with rollback"""
    try:
        conn.rollback()
        cur.executemany(sql, df.itertuples(index=False, name=None))
        conn.commit()
        print(f"✅ Table inserted successfully")
    except Exception as e:
        conn.rollback()
        print("❌ Insert failed:", e)

# --- Skills ---
skills_sql = """
INSERT INTO skills (skill_abr, skill_name)
VALUES (%s, %s)
ON CONFLICT (skill_abr) DO NOTHING;
"""
insert_table(skills_clean, skills_sql)

# --- Industries ---
industries_sql = """
INSERT INTO industries (industry_id, industry_name)
VALUES (%s, %s)
ON CONFLICT (industry_id) DO NOTHING;
"""
insert_table(industries_clean, industries_sql)

# =========================================
# Step 7: Insert bridge tables
# =========================================
# --- job_skills ---
job_skills_clean = job_skills.where(pd.notnull(job_skills), None)
job_skills_sql = """
INSERT INTO job_skills (job_id, skill_abr)
VALUES (%s, %s)
ON CONFLICT (job_id, skill_abr) DO NOTHING;
"""
insert_table(job_skills_clean, job_skills_sql)

# --- job_industries ---
job_industries_clean = job_industries.where(pd.notnull(job_industries), None)
job_industries_sql = """
INSERT INTO job_industries (job_id, industry_id)
VALUES (%s, %s)
ON CONFLICT (job_id, industry_id) DO NOTHING;
"""
insert_table(job_industries_clean, job_industries_sql)

cur.close()
conn.close()
print("✅ All tables inserted and connection closed")
```
